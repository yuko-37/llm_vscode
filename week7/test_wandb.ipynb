```python
import weave
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()
```

```python
weave.init('gpt-fine-tuning')
```

```python
client = OpenAI()
```

```python
@weave.op()
def create_completion(message):
    response = client.chat.completions.create(
        model='gpt-5',
        messages=[
            {'role': 'system', 'content': "You are are a helpful assistant"},
            {'role': 'user', 'content': message}
        ],
    )

    return response.choices[0].message.content
```

```python
message = 'Tell a lighthearted joke for the audience of LLM engineers.'
resp = create_completion(message)
```
